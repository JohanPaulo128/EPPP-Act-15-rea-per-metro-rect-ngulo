Se requiere desarrollar un programa en Python que funcione como una calculadora geométrica básica para estudiantes de preparatoria. El sistema debe presentar un menú interactivo que permita al usuario elegir entre calcular el área o el perímetro de un rectángulo, o bien salir del programa. Cada cálculo debe realizarse únicamente cuando el usuario seleccione la opción correspondiente, solicitando en ese momento las medidas necesarias. El programa debe mantenerse en ejecución hasta que el usuario decida terminarlo, mostrando los resultados de forma clara y amigable.
Formulas
1.  Círculo
*  2A = π × r²
*  P = 2 × π × r
2.  Rectángulo
*  A = base × altura
*  P = 2 × (base + altura)

```python
# Programa realizado por Johan Paulo Aros Reveles NC = 0016 NL = 4
# 📐 Calculadora Geométrica - Rectángulo 📏✨
# 🎓 Nivel: Principiante de Preparatoria
# Formulas a utilizar1.  Círculo
# Circulo
#2A = π × r²
# P = 2 × π × r
#  Rectángulo
#  A = base × altura
#  P = 2 × (base + altura)

# 🚫 Sin POO | 🚫 Sin validación | ✨ Menú interactivo

def area_rectangulo():
    """🟦 Calcula el área del rectángulo solicitando base y altura"""
    print("\n📐✨ ➡️ Calculando Área del Rectángulo... ✨📐")
    base = float(input("📏 Ingresa la base ➡️ "))
    altura = float(input("📏 Ingresa la altura ⬆️ "))
    area = base * altura
    print(f"✅🟦 ¡Listo! El área es: {area:.2f} unidades² 🎉")

def perimetro_rectangulo():
    """🔲 Calcula el perímetro del rectángulo solicitando base y altura"""
    print("\n📐🔄 ➡️ Calculando Perímetro del Rectángulo... 🔄📐")
    base = float(input("📏 Ingresa la base ➡️ "))
    altura = float(input("📏 Ingresa la altura ⬆️ "))
    perimetro = 2 * (base + altura)
    print(f"✅🔲 ¡Listo! El perímetro es: {perimetro:.2f} unidades 📏✨")

def menu():
    """📋 Muestra el menú y controla el bucle principal del programa"""
    print("\n🌟👋 ¡Bienvenido al Calculador Geométrico! 👋🌟")

    while True:
        print("\n📜📋  📖 MENÚ PRINCIPAL 📖  📋📜")
        print("1️⃣  🟦 Calcular área del rectángulo")
        print("2️⃣  🔲 Calcular perímetro del rectángulo")
        print("3️⃣  🛑 Salir del programa")

        opcion = input("\n👉 Selecciona una opción (1-3) ➡️ ")

        if opcion == "1":
            area_rectangulo()
        elif opcion == "2":
            perimetro_rectangulo()
        elif opcion == "3":
            print("\n👋💖 ¡Gracias por usar el programa! Hasta la próxima. ✨🚀")
            break
        else:
            print("⚠️🔍 Ops... Opción no válida. Por favor, elige 1️⃣, 2️⃣ o 3️⃣ 🔄")

# 🚀💻 Punto de entrada: ejecuta el programa
menu()
```
